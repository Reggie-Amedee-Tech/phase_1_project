## This provodes the central tendency and variablity of transportation metrics, including subway ridership and bridge traffic. It establishes a baseline for comparing pandemic and post pandemic behavior. 

```python
pandemic_mask = (df["date"] > "2020-03-12") & (df["date"] < "2023-03-06")
post_mask = df['date'] > "2023-03-06"

pandemic_data = df[pandemic_mask]
post_data = df[post_mask]
```

```python
pandemic_data.head()
```

```python
post_data.head()
```

```python
pandemic_data["subways_total_estimated_ridership"].mean()
```

```python
post_data["subways_total_estimated_ridership"].mean()
```

```python
pandemic_mean = pandemic_data["subways_total_estimated_ridership"].mean()
post_mean = post_data["subways_total_estimated_ridership"].mean()

print("Pandemic Mean:", pandemic_mean)
print("Post Mean:", post_mean)
```

```python
pandemic_std = pandemic_data["subways_total_estimated_ridership"].std()
post_std = post_data["subways_total_estimated_ridership"].std()

print("Pandemic STD:", pandemic_std)
print("Post STD:", post_std)
```

# Key Insight-- Percent Change (See Below)

```python
percent_change = ((post_mean - pandemic_mean) / pandemic_mean) * 100
print("Percent Change:", percent_change)
```

# Visuals that explore RIdership Recovery, Differences between periods and whether transit usage increased after the pandemic

```python
import matplotlib.pyplot as plt

periods = ["Pandemic", "Post-Pandemic"]
means = [pandemic_mean, post_mean]

plt.figure(figsize=(8,5))
plt.bar(periods, means)

plt.title("Average Subway Ridership: Pandemic vs Post-Pandemic")
plt.ylabel("Average Ridership")
plt.xlabel("Time Period")

plt.show()
```

# The average subway ridership during the pandemic period is significantly lower than in the post-pandemic period. This suggests a strong recovery in public transit usage after COVID-19 restrictions eased.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plt.plot(pandemic_data["date"], pandemic_data["subways_total_estimated_ridership"], label="Pandemic", alpha=0.6)
plt.plot(post_data["date"], post_data["subways_total_estimated_ridership"], label="Post-Pandemic", alpha=0.6)

plt.title("Subway Ridership Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Ridership")

plt.legend()
plt.xticks(rotation=45)

plt.show()
```

# This chart shows how subway ridership changed over time during the pandemic and post-pandemic periods. Ridership appears lower and more unstable during the pandemic, with a gradual increase in the post-pandemic period

```python
import matplotlib.pyplot as plt

data_to_plot = [
    pandemic_data["subways_total_estimated_ridership"],
    post_data["subways_total_estimated_ridership"]
]

plt.figure(figsize=(8,5))
plt.boxplot(data_to_plot, labels=["Pandemic", "Post-Pandemic"])

plt.title("Subway Ridership Distribution Comparison")
plt.ylabel("Ridership")

plt.show()
```

# This box plot compares the distribution of subway ridership during the pandemic and post-pandemic periods. The pandemic period shows lower and more spread-out ridership values, while the post-pandemic period shows higher and more stable usage overall.

```python
df_sorted = df.sort_values("date")

df_sorted["rolling_avg"] = df_sorted["subways_total_estimated_ridership"].rolling(window=7).mean()

plt.figure(figsize=(12,6))

plt.plot(df_sorted["date"], df_sorted["rolling_avg"])

plt.title("7-Day Rolling Average of Subway Ridership")
plt.xlabel("Date")
plt.ylabel("Ridership (7-day average)")
plt.xticks(rotation=45)

plt.show()
```

# This chart shows a 7-day rolling average of subway ridership, which smooths daily fluctuations and makes overall trends easier to see. It highlights the decline during the pandemic and gradual recovery afterward.

```python
df_sorted = df.sort_values("date")

df_sorted["pct_change"] = df_sorted["subways_total_estimated_ridership"].pct_change() * 100

plt.figure(figsize=(12,6))

plt.plot(df_sorted["date"], df_sorted["pct_change"])

plt.title("Daily Percent Change in Subway Ridership")
plt.xlabel("Date")
plt.ylabel("Percent Change (%)")
plt.xticks(rotation=45)

plt.axhline(0, linestyle="--")

plt.show()
```

# This chart shows the daily percent change in subway ridership, highlighting periods of sharp decline and recovery. The most significant negative changes occur during the pandemic period, while post-pandemic values fluctuate around smaller positive and negative changes, indicating stabilization in ridership trends.

```python
import matplotlib.pyplot as plt

df_sorted = df.sort_values("date")

df_sorted = df_sorted.set_index("date")

monthly = df_sorted["subways_total_estimated_ridership"].resample("ME").mean()

plt.figure(figsize=(12,6))

plt.plot(monthly.index, monthly.values)

plt.title("Monthly Average Subway Ridership")
plt.xlabel("Month")
plt.ylabel("Average Ridership")

plt.xticks(rotation=45)

plt.show()
```

# This chart shows monthly average subway ridership, which helps smooth daily fluctuations and makes the overall trend easier to see. It highlights a clear decline during the pandemic period followed by a gradual recovery in the post-pandemic period.
